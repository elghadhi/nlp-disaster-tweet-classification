**Converting text into numbers**

when dealing with a text problem, one of the first things you'll have to do before you can build a model is to convert text to numbers

There are a few ways to do this namely:

   .Tokenization: direct mapping of token(a token could be a word or a character) to number
   
   . Embedding: Create a matrix of feature vector(The Size of the feature vector can be defined and this embedding can be learned).



```python
import tensorflow as tf
from tensorflow.keras.layers import TextVectorization
```

```python
max_vocab_length =10000
max_length=15
text_vectorizer = TextVectorization(max_tokens=max_vocab_length,
                                    output_mode="int",
                                    output_sequence_length=max_length)
```

```python
#Fit (adapt) the text vectorizer to the training sentences
text_vectorizer.adapt(train_sentences)
```

```python
my_text = "There is a flood in my street "
text_vectorizer([my_text])
```

**Creating an Embedding using Embedding Layer**

```python
from tensorflow.keras import layers
```

```python
embedding = layers.Embedding(input_dim=max_vocab_length,
                             output_dim=128,
                             embeddings_initializer="uniform")
embedding
```

```python
text_vectorizer(["There is a flood in my street "])
```

```python
embedding(text_vectorizer(["There is a flood in my street "]))
```

```python
embedding(text_vectorizer(["There is a flood in my street "]))[0][0]
```

Nice, we have Converted texts into numbers

Model 1 LSTM

```python
from tensorflow.keras import layers

inputs = layers.Input(shape=(1,),dtype="string")
x=text_vectorizer(inputs)
x=embedding(x)

x=layers.LSTM(64)(x)

outputs = layers.Dense(1,activation="sigmoid")(x)

lstm_model = tf.keras.Model(inputs, outputs, name="LSTM_model")
```

```python
lstm_model.summary() # this is are the model architecture
```

```python
lstm_model.compile(loss="binary_crossentropy",
                   optimizer=tf.keras.optimizers.Adam(),
                   metrics=["accuracy"])
```

```python
lstm_model_history = lstm_model.fit(train_sentences,
                                    train_labels,
                                     epochs=5,
                                     validation_data=(val_sentences,val_labels))
```

```python
# Make predictions with LSTM model

lstm_model_pred_probs = lstm_model.predict(val_sentences)
lstm_model_pred_probs[:10]
```

```python
# Convert LSTM model probabilities to labels

lstm_model_preds = tf.squeeze(tf.round(lstm_model_pred_probs))
```

```python
lstm_model_preds[:10]
```

```python
from sklearn.metrics import accuracy_score, precision_recall_fscore_support

def calculate_result(y_True,y_pred):

  model_accuracy= accuracy_score(y_True,y_pred)*100
  model_precision , model_recal, model_f1, _= precision_recall_fscore_support(y_True,y_pred)

  model_results = {"accuracy": model_accuracy,
                   "Precision": model_precision,
                   "f1": model_f1}
  return model_results
```

```python
#Calculate model 3 results

lstm_model_result = calculate_result(y_True=val_labels,
                                    y_pred=lstm_model_preds.numpy())
lstm_model_result
```

**Model 2 : GRU**

```python
inputs = layers.Input(shape=(1,),dtype="string")

x=text_vectorizer(inputs) # tokenize The text

x= embedding(x) # Creates richer representation

x = layers.GRU(64)(x)

outputs = layers.Dense(1,activation="sigmoid")(x)
gru_model = tf.keras.Model(inputs,outputs,name="GRU_model")
```

```python
gru_model.summary()
```

```python
gru_model.compile(loss="binary_crossentropy",
                  optimizer=tf.keras.optimizers.Adam(),
                  metrics=["accuracy"])
```

```python
# Fit the model
gru_model_history = gru_model.fit(train_sentences,
                                  train_labels,
                                  epochs=5,
                                  validation_data=(val_sentences,val_labels))
```

```python
# Make Some predictions With Our GRU Model

gru_model_pred_probs = gru_model.predict(val_sentences)
gru_model_pred_probs
```

```python
val_labels
```

```python
# Convert GRU model pred probs to labels

gru_model_preds = tf.squeeze(tf.round(gru_model_pred_probs))
gru_model_preds
```

```python
# Calculate GRU model results

gru_model_results = calculate_result(y_True=val_labels,
                                      y_pred=gru_model_preds.numpy())
gru_model_results
```
